# Stack 자료구조 이해
## Stack의 개념과 특성
스택: 물건을 쌓아올리듯 자료를 쌓아올린 형태의 자료구조
대표적인 선형 자료구조 중 하나
### 스택의 특성
후입선출: 가장 마지막에 넣은 자료가 가장 먼저 나오는 것
1, 2, 3 => 3, 2, 1
### 스택의 기본연산
파이썬에서는 리스트를 사용해 구현할 수 있습니다.
저장소 자체를 스택이라 부르기도 함
스택에서 마지막 삽입된 원소의 위치: 스택 포인터, top으로 부르며 데이터를 넣거나 뺄 때 기준이 되는 위치입니다.
### 스택의 연산
삽입(Push): 저장소에 자료를 저장하는 연사으로, 보통 push라 부름
삭제(Pop): 저장소에서 삽입한 자료의 역순으로 꺼내는 연산으로 보통 pop이라고 부릅니다.
스택이 공백인지 아닌지를 확인하는 연산(isEmpty)
스택의 top에 있는 item(원소)을 반환하는 연산(peek)
### 스택의 삽입/삭제 과정
빈 스택에 원소 A,B,C를 차례로 삽입(push) 후 한번 삭제(pop)하는 연산 과정
### 스택 구현 시 고려 사항
스택의 크기 변경 어려움 => 동적 연결리스트

# Stack의 응용
## 괄호 검사
괄호의 조건: 
1. 왼쪽 괄호의 개수와 오른쪽 괄호의 개수가 같아야 한다.
2. 같은 괄호에서 왼쪽괄호는 오른쪽괄호보다 먼저 나와야 한다.
3. 괄호 사이에는 포함 관계만 존재한다.
## 스택을 이용한 괄호 검사
여는 괄호 push
닫는 괄호 stack에 남아있으면 pop
