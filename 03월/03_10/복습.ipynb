## 부분집합
- 완전탐색
  - 재귀호출을 이용한 완전탐색으로 부분집합을 구할 수 있다.
  - 실전 보다는 완전 탐색 학습용으로 추천하는 방법
- Binary Counting
  - 2진수와 비트연산을 이용하여 부분집합을 구할 수 있다.
  - 모든 부분 집합이 필요할 때 사용하는 추천 방법.

1. subset_recursion.py

```python
name = ['MIN', 'CO', 'TIM']

def recur(idx, subset):
    # 종료조건
    if idx == 3:
        print(subset)
        return
    # 재귀호출
    recur(idx+1, subset + [name[idx]])
    recur(idx+1, subset)

recur(0, [])
```

- 바이너리 카운팅(Binary Counting)
  - 원소 수에 해당하는 N개의 비트열을 이용해 부분집합을 표시
  - 001 이면 부분 집합 {A}를 나타냄
  - 0번 비트가 1이므로 첫 원소인 A만 포함된 부분집합을 나타낸다.
  - 110 이면 부분 집합 {B,C}를 나타냄
  - 1번,2번 비트가 1이므로, 두번째와 세번째 원소인 B,C가 포함된 부분집합을 나타낸다.
- 부분 집합의 총 개수
  - 만들 수 있는 집합의 총 개수는 2^n 이며,
    n = 3 이기에 총 8개의 부분 집합 존재
  - 2^n 은 1<<n 공식을 이용하여 빠르게 구할 수 있음.
  


2. subset_binary.py

```python
arr = ['a','b','c','d']
n = len(arr)

def get_subset():
    
    for i in range(1<<n):
        for idx in range(n):
            if i & 1<<idx :
                print(arr[idx], end=' ')
        print()

get_subset()
```

3. problem_go_cafe_with_friends.py
   5명중 2명 이상의 친구와 카페 가는 경우의 수

```python
# 바이너리카운팅버전
arr = ['A', 'B', 'C', 'D', 'E']
n = len(arr)


def get_count(tar):
    cnt = 0
    for i in range(n):
        if tar & 1<<i:
            cnt+=1
        
    return cnt
answer = 0
for target in range(1<<n):
    if get_count(target) >= 2:
        answer += 1
print(answer)
    
```

```python
# 재귀호출버전(내가짠코드)
arr = ['A', 'B', 'C', 'D', 'E']
n = len(arr)

answer = 0
def recur(idx,cnt):
    global answer
    # 종료조건
    if idx == 5 :
        if cnt >= 2:
            answer += 1
        return
    # 재귀호출
    recur(idx+1, cnt+1)
    recur(idx+1, cnt)

recur(0, 0)
print(answer)
```

4. combination recursion.py
   

```python
# 중복조합버전
# 5명 중 N명을 뽑을 것이다
arr = ['A', 'B', 'C', 'D', 'E']
N = 3
path = []

def combination(idx, start):
    # 종료조건
    if idx == 3:
        print(*path)
        return
    
    # 재귀호출
    for i in range(start,len(arr)):
        path.append(arr[i])
        combination(idx + 1, i)
        path.pop()

combination(0, 0)
```

```python
# 일반조합버전
arr = ['A', 'B', 'C', 'D', 'E']
N = 3
path = []

def combination(idx, start):
    # 종료조건
    if idx == 3:
        print(*path)
        return
    
    # 재귀호출
    for i in range(start,len(arr)):
        path.append(arr[i])
        combination(idx + 1, i+1)
        path.pop()

combination(0, 0)
```

```python
# 일반조합버전
arr = ['A', 'B', 'C', 'D', 'E']
N = 3
path = []

def combination(idx, prev):
    # 종료조건
    if idx == 3:
        print(*path)
        return
    
    # 재귀호출
    for i in range(prev+1,len(arr)):
        path.append(arr[i])
        combination(idx + 1, i)
        path.pop()

combination(0, -1)
```

5.dice.py

```python
N = 3
path = []

def dice(cnt, start):
    # 종료조건
    if cnt == 3:
        print(path)
        return
    # 재귀호출
    for i in range(start,7):
        path.append(i)
        dice(cnt+1, i)
        path.pop()

dice(0, 1)
```

```python
# subset 만들기로 연습
N = 3

def dice(cnt, start, subset):
    # 종료조건
    if cnt == 3:
        print(subset)
        return
    # 재귀호출
    for i in range(start,7):
        dice(cnt+1, i, subset + [i])


dice(0, 1, [])
```

# Greedy 탐욕알고리즘
- 결정이 필요할 때, 현재 기준으로 가장 좋아보이는 선택지로 결정하여 답을 도출하는 알고리즘
- 대표적인 문제해결 기법
  - 완전 탐색(Brute-Force)
    - 답이 될 수 있는 모든 경우를 시도해보는 알고리즘
  - Greedy
    - 결정이 필요할 때 가장 좋아보이는 선택지로 결정하는 알고리즘
  - DP
    - 현재에서 가장 좋아보이는 것을 선택하는 것이 아닌, 과거의 데이터를 이용하여 현재의 데이터를 만들어내는 문제해결 기법
  - 분할정복
    - 큰 문제를 작은 문제로 나누어 해결하는 문제해결 기법

6. problem_coin_swap.py

```python
# [문제] 동전의 최소 개수
# 규칙: 큰 동전부터 나누자
coin_list = [100, 50, 500, 10]
target = 1730
result = 0

coin_list.sort(reverse=True)

for coin in coin_list:
    possible_cnt = target // coin
    result += possible_cnt
    target = target - coin*possible_cnt

print(result)
```

- Greedy가 성립하는 경우 vs 성립하지 않는 경우
  - 10,50,100,500 처럼 모든 동전이 배수 관계인 경우는 Greedy 적용가능
  - 10,50,70 처럼 모든 동전이 배수 관게가 아닌 경우는 Greedy 적용불가
  - 그리디 알고리즘은 쉬워보이나, 예외없이 모든 경우가 맞는 규칙인지 아닌지 증명이 어렵다!
  

7. problem_toilet.py

```python
people = [15, 30, 50, 10]
n = len(people)

# 규칙: 최소시간인 사람부터 화장실로 들어가자.
people.sort()

total_time = 0
remain_people = n - 1

for turn in range(n):
    time = people[turn]
    total_time += time*remain_people
    remain_people -= 1

print(total_time)
```

8. problem_fractional_knapsack.py

```python
n = 3
target = 30
things = [(5, 50), (10, 60), (20, 140)]  # (Kg, Price)

# kg 당 가격으로 어떻게 정렬 ?
# 정렬 : (price / kg)
# lambda: 재사용하지 않는 함수
things.sort(key=lambda x: (x[1]/x[0]), reverse=True)

total = 0
for kg,price in things:
    per_price = price / kg

    # 만약 가방에 남은 용량이 얼마되지 않는다면,
    # 물건을 잘라 가방에 넣고 끝낸다
    if target < kg:
        total += per_price*target
        break

    total += price
    target -= kg

print(total)
```

## 연습문제
- 부분 집합의 합 문제
  - 아래의 10개의 정수 집합에 대한 모든 부분 집합 중 원소의 합이 0이 되는 부분집합을 모두 출력하시오.
  - {-1,3,-9,6,7,-6,1,5,4,-2}

```python
jiphop = [-1,3,-9,6,7,-6,1,5,4,-2]
N = len(jiphop)


def bubun(idx, subset):
    
    # 종료조건
    if idx == N:
        if sum(subset) == 0:
            print(subset)
        return
    # 재귀호출
    bubun(idx+1, subset + [jiphop[idx]])
    bubun(idx+1, subset)

bubun(0,[])
```
