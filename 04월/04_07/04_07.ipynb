2. V개의 정점이면, V-1개의 간선
3. 가중치의 합이 최소
- MST는 여러개가 가능

## Prim 
1. 큐에서 꺼낼건데 가장 작은것부터 먼저 꺼낸다=> 최소힙
## Kruskal 
1. 간선을 정렬
2. 사이클이 없으면 선택(=연결이 안되어있으면 선택)

## Prim
1. 인접리스트로 저장
2. 우선순위 큐(가중치,노드)
질문: 큐에 다넣지말고, 최솟값만 넣으면 안되는가?
검토: BFS에서 visited찍는타이밍과 Prim에서 찍는타이밍 차이
3. visited대신 MST

## Kruskal
질문: BFS,DFS,MST모두 모든 노드를 가는건가 (상황적이해필요)



## 어느 상황에서 Prim, Kruskal 쓰는가
- 시간복잡도:
- Kruskal => 정렬시간이 대부분 Elog(E)
- Prim => 전체간선확인(E) , 정점을 힙에 삽입/삭제 (logV) => Elog(V)
- 정점의 개수가 최대 1000개일때,최대간선의 개수는 999+998+...+1 = 50만개
- 밀집그래프이면 프림이유리, 희소그래프이면 크루스칼이 유리
- 질문: 무조건 프림이 유리한거아닌가?
