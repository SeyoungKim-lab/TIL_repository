## Django Template system
파이썬데이터를 HTML문서와 결합
1. 값 : python(딕셔너리 형태) => HTML( {{key}}형태 )
~~2. 필터 : {{ key|필터 }} (views.py에서 미리 수정해도됨)~~
3. 태그 : {%%}

## 순서리뷰
사용자 -> urls.py -> views.py의 함수(사용자 요청 처리) 호출 
-> 결과를 HTML응답

## 템플릿상속
dinner.html 상단에 {% extends "articles/base.htmal" %}
써주면, base.html의 {% block content %} 빈공간 {% endblock content %} 빈공간은 비워두고, 거기 들어갈 내용은 dinner.html에 작성.
**정리하면, dinner.html에 있는 내용이 base.html에 들어간다.**
