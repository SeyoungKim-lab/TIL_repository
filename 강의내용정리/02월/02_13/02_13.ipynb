 # 큐(Queue)
 먼저 들어온 데이터가 먼저 나가는 선형 자료구조
 놀이기구 줄
 선입선출: 가장 먼저 넣은 자료가 가장먼저 나오는것(FIFO)
 <-> 스택: 후입선출
 큐의 구조: 파이프(머리,꼬리)
 큐의 기본연산: | 스택:
 삽입: enqueue | push
 삭제: dequeue | pop
 # 큐의 연산과정
 1단계 공백 큐 생성: create_queue()
 front= rear= -1
 2단계 원소 A 삽입: enqeue(A)
 rear: 마지막(최근) 저장 원소
 rear는 +1 됨.
 3단계 원소 B 삽입: enqueue(B)
 4단계 원소 반환/삭제: dequeue()
 front가 +1됨
 5단계 원소 C 삽입: enqueue(C): rear +1
 6단계 원소반환/삭제 : dequeue():
 front +1
 7단계 원소반환/삭제: dequeue():
 front == rear

 # 선형 큐
 데이터를 일렬로 저장하며, 앞에서 꺼내고 뒤에 넣는 기본 큐 구조
 구현: 배열이나 연결 리스트로 구현할 수 있다.
       배열로 구현한 경우 큐의 크기는 배열의 크기와 같다.
       front: 가장 최근에 삭제된 원소의 인덱스
       rear: 마지막으로 저장된 원소의 인덱스
초기상태: front == rear = -1
공백상태: front == rear
포화상태: rear == n-1 (n: 배열의크기)

# 선형큐의 구현
1. 초기공백 큐 생성: create_queue()
크기 n인 1차원 배열 생성
front와 rear를 -1로 초기화
-create_queue()-
q = [0]*n
front = -1
rear = -1

2. 삽입 (enqueue(item))
   rear값을 하나 증가시켜 새로운 원소를 삽입할 자리를 지정
   그 인덱스에 해당하는 배열원소 q[rear]에 item을 저장

3. 삭제 dequeue()
   front += 1
   return q[front]
 or
q = []
q.pop(0)

4. 공백상태 및 포화상태 검사: is_empty() , is_full()
   return front == rear (True or False) : 공백상태

   return rear = len(q) - 1 :포화상태

5. 검색:qpeek()
   return q[front+1] : 가장 앞에 있는 원소를 검색하여 반환
   즉, 큐의 첫번째에 있는 원소를 반환
