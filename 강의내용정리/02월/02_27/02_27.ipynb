## UX(User Experience): 사용자의 만족도를 개선,최적화 위한 디자인과 개발분야
UX예시: 
1. 감각적 경험(브랜드인식): 러쉬 매장 근처만 가도 특유의향기, 개봉 시느껴지는 정교한 패키징의 손맛
2. 기능적경험(사용성): 음악 앱 검색창에 오타 => 찰떡같이 찾아주기
UX설계: 
이해 및 분석: 유저 리서치(인터뷰,설문), 데이터분석, 페르소나설정
구조화: 정보구조설계, 유저 시나리오 및 저니맵 작성
구체화및검증: 프로토타입

## UI : 서비스와 사용자 간의 상호작요을 가능하게 하는 디자인 요소들을 개발하고 구현하는 분야
UI예시: 
1. 물리적 UI: 리모컨
   자주쓰는버튼(전원,볼륨)의 크기와 위치 손에 쥐었을때의 그립감
2. 복합적 UI : ATM기기
   화면 속 터치버튼크기와 비치'
3. 디지털 UI: 웹/앱 로그인 화면
   로그인버튼의 위치, 색상, 붉은색경고문구 등
UI 설계:
핵심목표: 단순히 보기좋은 디자인이 아닌, 정보의 위계를 잡아 사용자가 무엇을 먼저 봐야할지 유도하고, 일관성을 통해 학습비용을 줄이는것

## Grid system 구조와 문법
1. Container : column들을 담고 있는 공간
2. Column : 실제 컨텐츠를 포함하는 부분
   Gutter : column 사이 여백
4. 1개의 row 안에 12개의 column영역이 구성
container => row => column
