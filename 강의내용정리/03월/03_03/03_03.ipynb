## 프로그래밍과 논리/수학
논리의 정의: 생각이나 주장의 타당성과 옳고 그름을 판별하는 규칙이나 원리를 의미.
경험적, 직관적, 확률적 판단을 의미하는 Soft logic에 반대되는 의미로 Hard logic을 의미.

Soft logic: 직관은 논리적인 느낌을 주는 것. 장점은 익숙한상황에서 빠르게 처리가능. 단점은 정확하지 않다.

명제: 
- 참이나 거짓을 알 수 있는 식이나 문장
- p,q,r로 표현
- 진릿값 : T,F 또는 1,0

명제의 연산(결합)
~p , p^q, p V q 
조건 명제: p → q 
역(뒤집기), 이(각각 not취하기) , 대우(not하고 뒤집기)
ex) p:비가 오면 q:우산을 쓴다.(참인 조건명제)
    역: 우산을쓰면 비가온다.(거짓)(오류가능성:비가 안올때도 우산을 쓸 수 있다.)
    이: 비가 오지않으면, 우산을 쓰지 않는다.(거짓)(오류가능성:비가 안올때도 우산을 쓸 수 있다.)
    대우: 우산을 쓰지 않으면 비가 오지 않는다.(참) - !중요!

증명: 명제를 증명할 수 있어야 한다.
증명 방법 여러가지
    - 증명: 논리적인 근거로 조건 명제가 참/거짓이다 라는 걸 말할 수 있다.  
1. 대우 증명(가장쉬움)
   - 4가 짝수라면, 4^2도 짝수이다.
   - 대우명제: 4^2이 홀수라면, 4는 홀수이다.
     - 가정이 거짓이므로 이 대우명제는 참이다.
     - 따라서 원레 명제도 참이다.
2. 직접 증명
   - 4가 짝수라면, 4^2도 짝수이다.
       - 4 = 2 (k)
       - 4^2 = (2k)^2 = 4k^2 = 2 (2k^2)
       - => 2로 나누어 떨어지는 수로 표현할 수 있으므로 짝수이다.
   - 귀납법( 대표적인 직접 증명 방식)
       - 특정 사례로부터 출발해서, 일반적인 규칙을 찾는 방법
       - 기저 사례
       - 귀납 가정
       - 귀납 단계
       - -> 하나가 되면 끝까지 된다.
      - 예시) 도미노
         - 명제: 모든 도미노는 연속으로 쓰러진다.
         - 기저 사례: 첫 번째 도미노를 손으로 밀면 쓰러진다.(P(1) 참인 걸 증명)
         - 귀납 가정: n번째 도미노가 쓰러졌다고 가정하자.
         - 귀납 단계: 도미노는 충분히 가깝게 배치되어 있어서, 하나가 쓰러지면 반드시 다음 것도 쓰러진다.
         - 따라서, n번째가 쓰러지면 => n+1 번째도 쓰러진다.
       - 예시2) [주의]경험적 귀납법 예시
         - 명제: n일 연속으로 하루 1번씩 더 연락하면, 상대의 호감도가 n단계 올라간다.
         - 기저 단계: 
           - 첫째 날, 평소보다 조금 더 연락 => 호감도 상승(1단계 상승)
         - 귀납 가정: n일 동안 매일 조금 더 연락하면, 호감도가 n 단계 오를 것이다!
         - 귀납 단계: 
           - 둘째 날 연락 조금더..
           - 셋째 날 연락 조금더..
           - n일 연속 연락 => n일 연속 호감도 상승?
