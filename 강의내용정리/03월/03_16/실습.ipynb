## 학습 개요

1. **학습 주제**
  - **Python 기본 문법**: Python의 기본 자료형(int, float, str, list, tuple, set, dict), 제어문(조건문/반복문), 함수 정의 및 활용
  - **객체지향 프로그래밍 기초**: 클래스(class) 개념, 객체/인스턴스/메서드의 역할, Python에서 모든 것이 객체인 이유
  - **데이터 분석 라이브러리**: NumPy를 활용한 배열 연산, Pandas를 활용한 DataFrame 생성 및 데이터 처리

2. **학습 목표**
 - Python의 다양한 자료형(int, float, str, list, dict, set)을 **선언하고 조작**할 수 있다.
 - 조건문과 반복문을 활용해 **흐름 제어 로직을 작성**할 수 있다.
 - 사용자 정의 함수와 클래스를 **구현**하고, 클래스가 왜 필요한지 **설명**할 수 있다.
 - NumPy 배열을 생성하고 **기본 수치 연산**을 수행할 수 있다.
 - Pandas DataFrame을 생성하고, 데이터를 **필터링·조회·분석**할 수 있다.
 - 각 라이브러리의 주요 함수가 **무엇을 반환하는지, 언제 사용하는지** 이해하고 활용할 수 있다.

3. **핵심 개념**
  - **기본 자료형 (int, float, str, None, bool)**: Python에서 숫자, 문자열, 불리언 값을 다루는 기본 타입입니다
  - **컬렉션 자료형 (list, tuple, set, dict)**: 여러 데이터를 저장하고 관리하는 자료구조로, 각각 순서·변경 가능 여부·중복 허용 여부가 다릅니다
  - **제어문 (if, for, while)**: 조건에 따라 코드 흐름을 제어하거나 반복 작업을 수행합니다
  - **함수 (function)**: 코드를 재사용 가능한 블록으로 만들어 가독성과 유지보수성을 높입니다
  - **클래스 (class)**: 데이터와 기능을 하나의 구조로 묶어 **재사용성과 구조화**를 제공하며, 객체/인스턴스/메서드 개념을 포함합니다
  - **NumPy 배열 (ndarray)**: 대규모 수치 데이터를 **벡터화 연산**으로 빠르게 처리하는 배열 기반 라이브러리입니다
  - **Pandas DataFrame/Series**: 표 형식(엑셀과 유사)의 데이터를 다루는 데 특화된 자료구조로, 데이터 조회·필터링·집계·변환을 쉽게 수행합니다
