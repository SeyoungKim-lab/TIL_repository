- Web Framework : 웹 어플리케이션을 빠르게 개발할 수 있도록 도와주는 도구 => 개발에 필요한 기본구조,규칙,라이브러리 등을 제공(로그인,로그아웃,회원관리,데이터베이스,보안 등) => 장고

- 가상환경
  - 하나의 컴퓨터 안에서 또다른 '독립된' 파이썬 환경
  - 하나의 집 안의 여러개의 방
  - 원레 pip에 인스톨하면 글로벌파이썬에 다운되게됨. 버전충돌가능성
  - 글로벌밑에 로컬파이썬(가상환경)

- 가상환경이 필요한 시나리오1
  - 한 개발자가 2개의 프로젝트를 진행해야 하는 상황
  - 프로젝트 A는 requests 패키지버전1이 필요
  - 프로젝트 B는 패키지버전2가 필요
  - 하지만 파이썬 환경에서 패키지는 1개의 버전만 존재가능
  - A와 B의 프로젝트의 다른 패키지 버전 사용을 위한 독립적인 개발 환경이 필요
  - 패키지란: 연관된 모듈들을 하나의 디렉토리에 모아놓은 것

- 1프로젝트에 1가상환경느낌
- 같은 집 안에 방을 따로 만들고 필요한물건을 그 방에만 들여놓음
- 방이 다르면 물건 달라도 간섭 X

- 가상 환경 생성 및 활성화 과정
  - 가상 환경 생성
    - python -m venv(가상환경 실행 명령어) venv(가상환경이름)
    - 가상환경이름은 바꿔도되지만 venv로 하기로 약속느낌
    - 혹은 앞에 쩜 붙여서 .venv => 앞에 쩜은 숨김을 의미
    - venv 폴더 안에는 파이썬 실행파일, 라이브러리 등을 담을공간마련됨

  - 가상 환경 활성화
    - source venv/Scripts/activate
    - 프로젝트A의 가상환경인지 B의 가상환경인지 직접 활성화해줘야함
    - 그게 위의 명령어임
    - 가상환경폴더venv내부의 스크립츠폴더 내부의 엑티베잇 이라는 파일이 있는데 이걸 직접 실행을 해줘야 가상환경이 실행됨
    - 1. 터미널 현재경로를 파악하기
      - 현재 경로위치를 PJT_A/venv로 이동해주고
      - 여기서 source venv/Scripts/activate 이거해주기(tab키 누르면 자동완성됨=>존재하는파일일때만 자동완성됨)
      - v 탭 s 탭 a 탭
    - 2. 가상 환경 종료
      - deactivate 하면 다시 글로벌환경으로 돌아옴(휴지통버튼 눌러도됨)
    - 1 터미널 1 가상환경
    - 1터미널가상환경키고 터미널 새로키면 그건 글로벌환경이됨.
    - 즉 하나의 터미널 안에서만 가상환경이 활성화된다!

- 의존성 패키지
  - 의존성 
    - 하나의 소프트웨어가 동작하기 위해 필요로 하는 다른 소프트웨어나 라이브러리
  - 의존성 패키지
    - 프로젝트가 의존하는 "개별 라이브러리"를 의미하는말
  - 패키지 목록 확인
    - 현재 가상환경에 설치된 라이브러리 목록을 확인하는 명령어
    - pip list
    - 기본은 pip와 setuptools 이렇게 두가지만 되어있음(파이썬 버전에 따라 다를 수 있다)

- 의존성 관리: 환경의 일치
  - 1. 의존성 기록(내보내기)
    - 현재 가상환경에 설치된 모든 패키지와 버전을 파일로 저장
    - pip freeze > requirements.txt
    - 설치된 패키지를 얼려서 텍스트파일로 옮기겠다.
    - 텍스트명은 관례적으로 사용.
  - 2. 의존성 설치(가져오기)
    - pip install -r requirments.txt
    - requirments.txt에 명시된 라이브러리를 한 번에 설치
  - 의존성 리스트 공유 시나리오
    - 2명의 개발자가 하나의 프로젝트를 함께 개발한다고 가정
    - 개발자 A가 먼저 가상환경을 생성후 프로젝트를 설정하고 관련된 패키지를 설치하고 개발하다가 협업을 위해 github에 프로젝트를 push
    - 개발자 B는 프로젝트를 clone받고실행해보려 하지만 실행되지않음
    - 개발자 A가 이 프로젝트르 위해 어떤 패키지를 설치햇고 어떤 버전을 설치햇는데 가상환경상황을 알수없음
    - 가상환경에 대한 정보 즉 패키지 목록이 공유되어야함

- 가상환경 주의사항 및 권장사항

  - 프로젝트마다 별도의 가상환경 사용

- Design Pattern
  - 소프트웨어 설계에서 반복적으로 발생하는 문제에 대한 검증되고 재사용 가능한 일반적인 해결책
  - 애플리케이션 구조는 이렇게 구성하자 라는 모범답안 또는 관행 대표적인 디자인패턴: MVC

- MVC 디자인 패턴
  - 하나의 애플리케이션을 구조화하는 대표적인 구조적 디자인 패턴
    - Model: 데이터 및 비즈니스 로직을 처리
    - View: 사용자에게 보이는 화면을 담당
    - Controller: 사용자의 입력을 받아 모델과 뷰를 제어.
  - 시각적 요소와 뒤에서 실행되는 로직을 서로 영향없이, 독립적이고 쉽게 유지 보술하 ㄹ수 있는 애플리케이션을 만들기 위함

- MTV 디자인 패턴
  - 장고에서 애플리케이션을 구조화하는 디자인 패턴
  - View를 Template 으로, Controller를 View로 명칭을 바꿈
  - 장고에서의 뷰는 컨트롤러를 의미. 시각적인 영역이 템플릿.

- 프로젝트와 앱
  - Django에서 프로젝트와 앱의 관계
  - 프로젝트 안에 앱A, 앱B가 있다.
  - 프로젝트 안에는 로그인,로그아웃,회원가입,게시글작성,게시글삭제 등과 같은 기능이 있음.
  - 얘네들을 기능별 목적에 맞게끔 카테고리화 해서 넣음. 이걸 각각 app.
  - 사용자관리app, 게시글관련app 등 기능별목적.

- 프로젝트: 앱의 집합
- 앱: 독립적 작동하는 기능단위 모듈
- 앱의 분류기준은 개발자마다 다름.

- 장고의 요청과 응답
  - urls.py 는 카페의 메뉴판 느낌. 그 메뉴판에 요청한 메뉴가 있어야만 제공가능.
  - 요청이 들어오면 urls.py을 열고 그 안에 그 요청이 들어있다면..
  - view.py는 그 메뉴를 만들고 model과 templates과 상호작용하여 사용자에게 내보냄
- articles 앱 폴더 안에 templetes 폴더 생성하고 그 안에 articles폴더 생성 그 안에 템플릿 파일 생성
