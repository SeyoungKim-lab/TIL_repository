## Django Template system
파이썬데이터를 HTML문서와 결합
1. 값 : python(딕셔너리 형태) => HTML( {{key}}형태 )
~~2. 필터 : {{ key|필터 }} (views.py에서 미리 수정해도됨)~~
3. 태그 : {%%}

## 순서리뷰
사용자 -> urls.py -> views.py의 함수(사용자 요청 처리) 호출 
-> 결과를 HTML응답

# Django Model 복습
- urls.py : 사용자 요청의 시작점
- views.py: 요청을 처리하고 models.py를 통해 데이터를 다룸
- models.py: 데이터베이스를 정의하고, 데이터베이스와 상호작용
- templates: views.py 로부터 받은 데이터를 사용자에게 보여줄 화면을 구성
## Model class
- Model class는 데이터베이스 테이블의 구조를 설계하는 '청사진' 역할을 합니다.
- 어떤 데이터(컬럼)을 저장할지, 그 데이터의 형태는 어떠할지(타입,길이)를 Python코드로 명확히 정의

## 앱마다 url을 따로 지정
- urls.py 에서 from django.urls import path,include 해주고,
- path('articles/', include('articles.urls') )
- articles 폴더의 urls.py 에서 from django.urls import path, from . import views 넣어주고
- app_name = 'articles' , urlpatterns = []

## models.py
- from django.db import models , 
- class Article(models.Model):      django.db.models 모듈의 Model 클래스를 상속하겠다.
    - title = models.CharField(max_length=10)   
    - content = models.TextField()
- 수정 후 python manage.py makemigration 하고 마이그레이트까지

## Model Field
- DB 테이블의 필드(열) 정의
- 데이터 타입 및 제약 조건 명시
- models 모듈의 클래스로 정의되어 있음
- CharField() : 제한된 길이의 문자열을 저장, max_length 는 선택옵션
- TextField() : 길이 제한이 없는 대용량 텍스트를 저장
- 문자열필드: CharField, TextField
- 숫자필드: IntegerField, FloatField
- 날짜/시간필드: DateField, TimeField, DateTimeField
- 파일관련필드: FileField, ImageField
- 구글에 django model field 검색 -> Field Type
### Field options
- 제약조건
  - 특정 규칙을 강제하기 위해 테이블의 열이나 행에 적용되는 규칙이나 제한사항
