```python
import sys
sys.stdin = open("input_folder/최대상금.txt", "r")

T = int(sys.stdin.readline())
 
# now_num: 현재 숫자
# cnt: 현재까지 바꾼횟수
def DFS(cnt):
    global max_v
 
    
    # 종료조건
    if cnt == opertunity:
        number = int(''.join(list_num))
        max_v = max(number, max_v)
        return
     
    # 탐색(재귀)
    for i in range(len(list_num)):
        for j in range(i+1,len(list_num)):
            list_num[i], list_num[j] = list_num[j], list_num[i] # 원본건듦
            next_list = list_num
            next_num = int(''.join(next_list))
            # (다음숫자,다음깊이)를 이미 방문했다면 컨티뉴
            if (next_num, cnt+1) in visited:
                list_num[i], list_num[j] = list_num[j], list_num[i] # 원상복구 ..
                continue
            visited.add((next_num, cnt+1))
            DFS(cnt+1) 
            list_num[i], list_num[j] = list_num[j], list_num[i] # 원상복구
 
 
for tc in range(1,1+T):
    number, opertunity = map(int, sys.stdin.readline().split())
    list_num = list(str(number))
     
    visited = set()
    max_v = 0
 
    visited.add((number, 0))
    DFS(0)
 
    print(f"#{tc} {max_v}")
```
